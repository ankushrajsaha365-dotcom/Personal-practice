```python
#vowel or consonant

word = input("Enter a word:")
s = word.lower()
for i in s:
    if i == 'a' or i == 'e' or i == 'i' or i == 'o' or i == 'u':
       print(f"{i} is a vowel")
    else:
        print(f"{i} is a consonant")
```

```python
#remove characters except alphabets
l1=[]
l2=[]
l3=[]
string = input("Enter a string:")
for i in string:
    c = ord(i)
    l1.append(c)
for j in l1:
    if j in range(97,123):
        l2.append(j)
    if j in range(65,91):
        l2.append(j)
# print(l2)
for e in l2:
    w = chr(int(e))
    l3.append(w)
# print(l3)
result = ''.join(l3)
print(result)
```
