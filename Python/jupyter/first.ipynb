```python
import matplotlib.pyplot as plt

# Data
Date = ["25/12", "26/12", "27/12", "28/12", "29/12"]
temp = [8.5, 10.5, 6.8, 6.9, 5.2]

# Plotting the line chart
plt.plot(Date, temp, marker='o')

# Adding labels
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")

# Adding title
plt.title("Temperature Trend Over Dates")

# Adding grid
plt.grid(True)

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Corrected data
height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

# Plotting the line chart
plt.plot(weight, height, color='green', linestyle='--', linewidth=2)

# Axis labels
plt.xlabel("Weight in kg")
plt.ylabel("Height in cm")

# Title
plt.title("Average weight with respect to average height")

# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
course = ["C", "C++", "Java", "Python"]
students = [20, 15, 30, 35]   # number of students (your 'weight' list)

# Bar Plot
plt.figure(figsize=(8,5))
plt.bar(course, students, color="maroon")

# Scatter Plot
plt.scatter(course, students, color="blue")

# Labels & Title
plt.xlabel("Courses offered")
plt.ylabel("No. of students enrolled")
plt.title("Students enrolled in different courses")

# Show Plot
plt.show()
```
