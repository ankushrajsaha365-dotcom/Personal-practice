Consider the average heights and weights of persons aged 8 to 6 stored in the following two lists:
height = [121.9,124.5,129.5,134.6,139.7,147.3,152.4, 157.5. 162.6]
weight= [19.7,21.3,23.5.25.9,28.5.32. 1,35.7,39.6,43.2]
Let us plot a line chart where:
I. X axis will represent weight.
II. Y axis will represent height.
III. X axis label should be &quot;Weight in kg”.
IV. Y axis label should be &quot;Height in cm&quot;.
V. Colour of the line should be green.
VI. The title of the chart should be *&quot;Average weight with respect to average height&quot;.
VII. Line style should be dashed.
VIII. Linewidth should be 2.

```python
import matplotlib.pyplot as plt

height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

plt.plot(weight, height, color='green', linestyle='--', linewidth=2)

plt.xlabel("Weight in kg")
plt.ylabel("Height in cm")
plt.title("Average weight with respect to average height")

plt.show()
```

```python
course = np.array(["C", "C++", "Java"," Python"])
weight= np.array([20,15,30,35])

plt.subplot(2,1,1)
plt.bar(course,weight)
plt.xlabel("Course offered")
plt.ylabel("No. of students enrolled")
plt.yticks(rotation=45)

plt.subplot(2,1,2)
plt.scatter(course,weight)
plt.xlabel("Course offered")
plt.ylabel("No. of students enrolled")
plt.yticks(rotation=45)

plt.tight_layout()
plt.show()
```
