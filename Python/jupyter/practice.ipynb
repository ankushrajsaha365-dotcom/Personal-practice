>>Write a python program to swap two variables using and without using third variable.

```python
def swap1(x,y):
    x,y = y,x
    print("After swaping 'a' is:",x)
    print("After swaping 'b' is:",y)

def swap2(x,y):

    temp = x
    x = y
    y = temp

    print("After swaping 'a' is:",x)
    print("After swaping 'b' is:",y)



a = int(input("Enter a number 'a':"))
b = int(input("Enter a number 'b':"))

print(f"Before swaping a = {a} and b = {b}")

ch = int(input("Enter 1 to swap without 3rd variable\nEnter 2 to swap using 3rd variable"))

if 1<=ch<=2:
    if ch == 1:
        swap1(a,b)
    else:
        swap2(a,b)

else:
    print("Wrong choice!!!")

```

>>Write a python to generate Fibonacci Series up-to n-terms using loop.

```python
def term(num):
    a,b = 0,1
    print("The Fibonacci Series is(upto n terms):")
    for i in range(num):
        print(a,end=" ")
        a,b = b,a+b
def limit(num):
    a,b =0,1
    print("\nThe Fibonacci Series is(upto limit):")
    while a<=num:
        print(a,end=" ")
        a,b = b,a+b

n = int(input("Enter the number:"))


term(n)
limit(n)
```

>>Write a python program to generate all PrimeNumbers within a range, where range is userinput.

```python
n = int(input("Enter the range: "))

for i in range(2, n):
    is_prime = True

    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            is_prime = False
            break

    if is_prime:
        print(i, "is Prime")
    else:
        print(i, "is Not Prime")
```

>>Write a python program to reverse a number and check whether it is a Palindrome.

```python
num = int(input("Enter a number: "))

# Store original number
original = num

# Reverse the number
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10

print("Reversed number:", rev)

# Check palindrome
if original == rev:
    print("It is a Palindrome.")
else:
    print("It is not a Palindrome.")
```

>>Write a python program to check whether a number is Armstrong number or not.

```python
num = int(input("Enter a number: "))

# Count number of digits
n = len(str(num))

temp = num
sum_digits = 0

while temp > 0:
    digit = temp % 10
    sum_digits += digit ** n
    temp //= 10

if sum_digits == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is NOT an Armstrong number")
```

>>An automorphic number is the number which contained in last digit(s) of its square. Example 76 is an automorphic number as its square is 6776 and 76 is present as the last two digits. Write a python script to check whether a number is automorphic number.

>>A number is said to be a Krishnamurthy number, if thesum of the factorial of the digits of a number is same as the original number. Example-145 is a special number, because 1! + 4! + 5! =145. Write a python program to check whether a number is Krishnamurthy Number or not.

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n*fact(n-1)

n=int(input("Enter a number:"))
sum = 0
temp = n

while temp>0:
    digit = temp%10
    temp//=10
    f = fact(digit)
    sum+=f
if sum == n:
    print("Krishnamurthy Number")
else:
    print("Not a Krishnamurthy numbeer")
```

 Write a Python program to print Tribonacci Series (1, 1, 2, 4, 7, 13, 24, ……….)
using user defined function.

```python
#Tribonacci Series
def trib(num):
    a,b,c = 0,1,1
    for i in range(num):
        print(a,end=" ")
        a,b,c = b,c,a+b+c
        
#Fibonacci Series
def fib(num):
    a,b = 0,1
    for i in range(num):
        print(a,end=" ")
        a,b = b,a+b
        

num = int(input("Enter range:"))

print("The Fibonacci Series:")
fib(num)
print("\n")
print("The Tribonacci Series:")
trib(num)
```
