````python
class A:
    def get_num(self, numbers):
        self.numbers = numbers

    def get_mean(self):
        return sum(self.numbers) / len(self.numbers)


class B:
    def get_median(self):
        nums = sorted(self.numbers)
        n = len(nums)
        mid = n // 2

        if n % 2 != 0:
            return nums[mid]
        else:
            return (nums[mid - 1] + nums[mid]) / 2


class C(A, B):
    def get_mode(self):
        return max(self.numbers, key=self.numbers.count)


# -------- Main Program --------
numbers = list(map(int, input().split()))

formula = C()
formula.get_num(numbers)

print("Mean:", formula.get_mean())
print("Median:", formula.get_median())
print("Mode:", formula.get_mode())
````
