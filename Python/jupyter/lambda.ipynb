A lambda function is a small anonymous function

A lambda function can take any number of arguments, but can only have one
expression

```python

f = lambda num: num*2

num = int(input())
print(f(num))
```

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x*2, numbers))
print(result)
```

```python
x = lambda a : a + 10

print(x(5))
```

```python
x = lambda a, b : a * b

print(x(5, 6))
```

```python
x = lambda a, b, c : a + b + c

print(x(5, 6, 2))
```

```python
def myfunc(n):
    return lambda a : a * n

mydoubler = myfunc(2)

print(mydoubler(11))
```

Write a Python program to add two numbers using lambda function

Write a Python program to find smaller of two numbers using lambda
functions

```python
def small(a, b):
    return (lambda x, y: x if x < y else y)(a, b)

num1 = int(input())
num2 = int(input())

print(small(num1, num2))
```

```python
def small(a, b):
    if a < b:
        return (lambda x: x)(a)
    elif a > b:
        return (lambda x: x)(b)
    else:
        return "Both numbers are same!"

num1 = int(input())
num2 = int(input())

print(small(num1, num2))
```
