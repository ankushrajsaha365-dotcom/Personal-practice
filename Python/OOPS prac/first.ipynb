```python
# Area of Rectangle

class rectangle:
    def Area(self,a,b):
        return a*b

r = rectangle()
a = int(input())
b = int(input())
print(r.Area(a,b))
```

```python
# Area and perimeter of a circle

class Circle:
    def area(self,r):
        return 3.14*r*r
    def perimeter(self,r):
        return 2*3.14*r

c = Circle()
r = float(input())
print("Area:",c.area(r))
print("Perimeter:",c.perimeter(r))
```

```python
# simple claculator

class Calculator:
    def input(self):
        self.x = float(input("Enter a number x:"))
        self.y = float(input("Enter a number y:"))

    def add(self):
        print("Addition:",self.x+self.y)

    def sub(self):
        print("Substraction:",self.x-self.y)

    def mul(self):
        print("Multiplication:",self.x*self.y)

    def div(self):
        print("Division:",self.x/self.y)

c = Calculator()

f = True
while f:
    print("-----------Menu------------\n1->Addition\n2->Substraction\n3->Multiplication\n4->Division\n")
    choice = int(input("Enter your choice:"))
    
    if choice == 1:
        c.input()
        c.add()
    elif choice == 2:
        c.input()
        c.sub()
    elif choice == 3:
        c.input()
        c.mul()
    elif choice == 4:
        c.input()
        c.div()
    else:
        f = False    
```

```python
### Even or Odd
class ev_odd:
    def input(self):
        self.a = int(input("Enter a number :"))
    def check(self):
        if self.a % 2 == 0:
            print("Even")
        else:
            print("Odd")

num = ev_odd()
num.input()
num.check()
```

```python
### Result

class Student:
    def subject(self):
        self.a = int(input("Number of sub A:"))
        self.b = int(input("Number of sub B:"))
        self.c = int(input("Number of sub C:"))

    def check(self):
        self.total = (self.a+self.b+self.c)


print("------STREAM------\n101->Science\n102->Commerce\n102->Arts\n")
stream = int(input("Enter your stream ID:"))

student = Student()

if stream in [101,102,103]:
    student.subject()
    student.check()
    if stream == 101:
        if student.total >=260:
            print("Pass")
        else:
            print("F")
    elif stream == 102:
        if student.total >=240:
            print("Pass")
        else:
            print("F")
    elif stream == 103:
        if student.total >=220:
            print("Pass")
        else:
            print("F")
else:
    print("Invalid ID!!!")
```
