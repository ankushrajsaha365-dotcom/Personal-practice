```python
class Class1:
    def m(self):
        print("In Class1")

class Class2(Class1):
    def m(self):
        print("In Class2")
        Class1.m(self)

class Class3(Class1):
    def m(self):
        print("In Class3")
        Class1.m(self)

class Class4(Class2, Class3):
    def m(self):
        print("In Class4")
        Class2.m(self)
        Class3.m(self)

obj = Class4()
obj.m()
```

```python
class Class1:
    def m(self):
        print("In Class1")

class Class2(Class1):

    def m(self):
        print("In Class2")
        super().m()

class Class3(Class1):
    def m(self):
        print("In Class3")
        super().m()

class Class4(Class2, Class3):
    def m(self):
        print("In Class4")
        super().m()

obj=Class4()

print(obj.m())
```

```python
class ABC:
    pass

from abc import ABC
class Polygon(ABC):
    def psides(self):
        pass

class Triangle(Polygon):
    def psides(self):
        print("Triangle has 3 sides")

class Rectangle(Polygon):
    def psides(self):
        print("Rectangle has 4 sides")

R = Triangle()

R.psides()

K = Rectangle()

K.psides()
```
