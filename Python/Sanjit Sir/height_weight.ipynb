```python

df = pd.read_csv('height_weight.csv')

x = df['Weight']
y = df['Height']

n = len(x)

# Means
x_mean = x.mean()
y_mean = y.mean()

# Slope m
m = sum((x - x_mean) * (y - y_mean)) / sum((x - x_mean)**2)

# Intercept b
b = y_mean - m * x_mean

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
y_pred = m*x + b

plt.scatter(x, y)
plt.plot(x, y_pred, color='red')
plt.xlabel("Student Weight")
plt.ylabel("Contact Height")
plt.title("Student Height/Weight")
plt.show()
```
