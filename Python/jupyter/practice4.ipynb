Write a Python user define function rotatelist(l,k) that takes a list l and a positive integer k and
returns the list l after k rotations. If k is not positive, your function should return l unchanged. Note
that your function should not change itself, and should return the rotated list. A list rotation
consists of taking the last element and moving it to the front. For instance, if we rotate the list
[1,2,3,4,5], we get [5,1,2,3,4]. If we rotate it again, we get [4,5,1,2,3].

Here is an examples to show how your function should work:

Input:
rotatelist([1,2,3,4,5],3)

Output: [3, 4, 5, 1, 2]

```python
def rotatelist(l):
    last = l.pop()
    l.insert(0,last)

l1 = []
n = int(input("Enter number of elements:"))
for i in range(n):
    e = int(input("Enter element:"))
    l1.append(e)
print("Original list:",l1)
l2 =l1.copy()
rot = int(input("How many rotations you want to perform:"))
for j in range(rot):
    rotatelist(l2)
print("New list:",l2)
```

Write a Python function that checks whether a word or phrase is palindrome or not.

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g.,
madam, kayak, racecar, or a phrase "nurses run".

```python
def palindrome(str):
    str = str.lower()
    if str == str[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")

str = input("Enter a word")
palindrome(str)
```

Write a Python function to check whether a string is pangram or not. (Assume the string passed in
does not have any punctuation).

Note : Pangrams are words or sentences containing every letter of the alphabet at least once.

For example : "The quick brown fox jumps over the lazy dog"

Hint: You may want to use .replace() method to get rid of spaces.

```python
def pangram(str):
    str = str.replace(' ','')
    char = 'abcdefghijklmnopqrstuvwxyz'
    str = str.lower()
    for ch in char:
        if ch not in str:
            return False
    return True
str = input("Enter your sentence:")
if pangram(str) == True:
    print("Pangram")
else:
    print("Not Pangram")
```

Write a function that checks whether a number is in a given range (inclusive of high
and low).

```python
def range(l,u,num):
    if l<=num<=u:
        print("Yes , the number is in the range")
    else:
        print("No, the number is not in the given range")


up = int(input("Enter the upper range:"))
low = int(input("Enter the lower range:"))

n = int(input("Enter a number:"))

range(low,up,n)
```

Write a Python function that accepts a string and calculates the number of upper case
letters and lower case letters.

Sample String: 'Hello Mr. Rogers, how are you this fine Tuesday?'

Expected Output:
No. of Upper case characters : 4

No. of Lower case Characters : 33

```python
def upper(str):
    up_count = 0
    for i in str:
        if 65<=ord(i)<=90:
            up_count+=1
    return up_count
def lower(str):
    low_count = 0
    for i in str:
        if 97<=ord(i)<=122:
            low_count+=1
    return low_count

str = input("Enter the sentence:")



up = upper(str)
low = lower(str)

print(f"No. of Upper case characters:{up}\nNo. of Lower case characters:{low}")
```

Write a user define Python Program to check whether a positive number is as SUM
of SQUAREs or NOT. A positive integer m is a sum of squares if it can be written as
k + l where k > 0, l > 0 and both k and l are perfect squares.

Here are some examples to show how your function should work:

sumofsquares(25)  25 is a SUM of SQUAREs,

sumofsquares(36)  25 is NOT a SUM of SQUAREs,

sumofsquares(225)  25 is a SUM of SQUAREs

```python
def sumsq(num):
    root = int(num**0.5)
    for i in range(1,root+1):
        for j in range(1,root+1):
            if i*i + j*j == num:
                return True
            
    return False
            
n = int(input("Enter a number:"))

if sumsq(n) == True:
    print("The given number is a sum of squares")
else:
    print("The given number is not a sum of squares")

```
