1. Write a Python program to print date and time using time
module and print the calendar of a particular month

```python
# from datetime import datetime , date , timedelta
# print(datetime.now())

import calendar

year = int(input("Enter year:"))
month = int(input("Enter month:"))
if 1<=month<=12 and year>0:
    print(calendar.month(year,month))
else:
    print("Wrong input!!!")
```

2. Write a Python program to print absolute value, square root and cube of a number using math module

```python
import mymath

n = int(input("Enter a number:"))

#absolute number

print("Absolute:", mymath.abs(n))

#square root

print("Square root:", mymath.sq_root(n))

#cube

print("Cube:", mymath.cube(n))
```

3. Define a factorial function in a user-defined module
“Fact” and use command line argument to find the factorial
value of the input

```python
import fact

n = int(input("Enter a number:"))

print("Factorial is:",fact.fact(n))
```
