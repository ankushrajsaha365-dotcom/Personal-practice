```python
import pandas as pd

# Series (like a single column)
s = pd.Series([10, 20, 30, 40])
print(s)

# DataFrame (like a table)
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["Delhi", "Mumbai", "Kolkata"]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df[df["Age"] > 28])  # all rows where Age > 28
```

```python
print(df.loc[df["Age"] > 28, ["Name", "City"]])
```

```python
print(df.sort_values(by="Age" , ascending=False))
```

```python
print(df["Age"].mean().astype(int))
print(df["Age"].max())
print(df["Age"].min())
```
