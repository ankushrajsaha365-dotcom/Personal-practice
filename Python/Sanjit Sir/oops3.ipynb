```python
class Mean:
    def __init__(self):
        self.numlist = []

    def get_num(self):
        size = int(input("Enter list size:"))
        for i in range(size):
            num = int(input("Enter elements:"))
            self.numlist.append(num)
    def mean_num(self):
        sum = 0
        for i in self.numlist:
            sum +=i
        mean = sum/len(self.numlist)
        return mean
        
    
meanvalue = Mean()
meanvalue.get_num()
result = meanvalue.mean_num()
print("Mean:",result)

```

```python
class Complex:
    def __init__(self,real,imaginary):
        self.real = real
        self.imaginary = imaginary

    def putData(self):
        print(self.real,"+",self.imaginary,"i")
    
def __main__():
    c1 = Complex(5,3)
    c1.putData()

__main__()
```
