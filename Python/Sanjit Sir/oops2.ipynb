```python
# Single Inheritance
class Person:
    def __init__(self,name,id):
        self.name = name
        self.id = id
    def get_details(self):
        return f"Name : {self.name} , ID : {self.id}"
    
class Student(Person):
    def __init__(self,name,id,grade,course):
        super().__init__(name,id)
        self.grade = grade
        self.course = course
    def get_details(self):
        return f"Name : {self.name} , ID : {self.id} , Grade : {self.grade} , Course : {self.course}"
    

stud = Student("ABC",123,"O",["CSE","DATASCIENCE","AIML"])
print(stud.get_details())
```

```python
# Multiple Inheritance
class Person:
    def get_details(self):
        return "Details of a person"
class Athlete:
    def get_skill(self):
        return "Athletic skill"
    
class Student(Person,Athlete):
    pass

student = Student()
print(student.get_details())
print(student.get_skill())
```

```python
# multiple inheritance application
# define two classes Circle and Cube and access them from a different class to calculate 
# the area of the circle and the volume of the cube dependent on the choice of user

## ver. 1.0
class Circle:
    def __init__(self):
        self.radius = float(input("Enter the radius:"))
        self.area = self.Area()
    def Area(self):
        self.area = 3.14*pow(self.radius,2)
        print(f"Area is {self.area}")

class Cube:
    def __init__(self):
        self.length = float(input("Enter the length of each side of the cube:"))
    def volume(self):
        self.vol = pow(self.length,3)
        print(f"Volume is {self.volume}")


class Math(Circle,Cube):
    def cal(self):
        print("1: Radius of circle\n2: Vol of cube\n************\n")
        
        self.choice = int(input("Enter your choice:"))
        if math.choice == 1:
            math.Area()
        elif math.choice == 2:
            math.volume()
        else:
            print("Wrong choice!!!")

math = Math()
math.cal()


```

```python
## Multilevel Inheritance

class Person:
    def __init__(self,name,id):
        self.name = name
        self.id = id
    def get_details(self):
        return f"Name : {self.name} , ID : {self.id}"
    
class Student(Person):
    def __init__(self,name,id,grade,course):
        super().__init__(name,id)
        self.grade = grade
        self.course = course
    def get_details(self):
        return f"Name : {self.name} , ID : {self.id} , Grade : {self.grade} , Course : {self.course}"
    
class Project(Student):
    def __init__(self,name,id,grade,course,p_title):
        super().__init__(name,id,grade,course)
        self.p_title = p_title
      
    def get_details(self):
        return f"Name : {self.name} , ID : {self.id} , Grade : {self.grade} , Course : {self.course} , Project Title :{self.p_title}"
    

project = Project("ABC",123,"O","AIML","XYZ")
print(project.get_details())
```
