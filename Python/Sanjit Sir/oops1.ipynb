```python
class rectangle:
    def __init__(self):
        self.length = int(input("Enter length: "))
        self.breadth = int(input("Enter breadth: "))    
        area = self.length * self.breadth
        print("Area of rectangle is:", area)

obj = rectangle()
```

```python
# using two constructors    
# constructors cannot be overloaded in python

class rect:
    def __init__(self):
        self.length = int(input("Enter length: "))
        self.breadth = int(input("Enter breadth: "))    
    def __init__(self,length,breadth):
        area = self.length*self.breadth
        print("Area of rectangle is:", area)

obj = rect(2,4)
```

```python
###

class Book:
    def getddata(self):
        self.Name = input("Enter name of book: ")
        self.Publisher = input("Enter publisher of book: ")
        self.Year = int(input("Enter year of book: "))
        self.ISBN = int(input("Enter ISBN of book: "))
        self.Authors = input("Enter authors of book: ")
        
    def putdata(self):
        print("Name:",self.Name)
        print("Publisher:",self.Publisher)
        print("Year:",self.Year)
        print("ISBN:",self.ISBN)
        print("Authors:",self.Authors)

book = Book()
book.getddata()
book.putdata()
```
