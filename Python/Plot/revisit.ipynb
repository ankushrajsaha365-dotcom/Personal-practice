```python
# Scatter plot

import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,6,9,5,6,7,8])
y = np.array([99,86,87,88,100,86,103,87,94,78])

plt.scatter(x,y, color = "orange" ,marker = "h" , s = 100, alpha = 0.7)
plt.show()
```

```python
#bar charts

x = np.array(["A","B","C","D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color="purple", width=0.5)
plt.show()
```

```python
# Histogram

data = np.random.normal(50, 10, 200)  
# 200 random numbers, mean=50, std=10

plt.hist(data, bins=20, color="skyblue", edgecolor="black")
plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np

# two groups of scores
classA = np.random.normal(70, 10, 200)   # mean=70, std=10
classB = np.random.normal(65, 15, 200)   # mean=65, std=15

plt.hist(classA, bins=20, alpha=0.6, label="Class A", color="blue", edgecolor="black")
plt.hist(classB, bins=20, alpha=0.6, label="Class B", color="red", edgecolor="black")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

classA = np.random.normal(70, 10, 200)
classB = np.random.normal(65, 15, 200)

plt.hist([classA, classB], bins=20, 
         label=["Class A", "Class B"], 
         color=["blue", "red"], 
         edgecolor="black", 
         histtype="bar")

plt.legend()
plt.show()
```

```python
#Pie Chart

values = np.array([35, 25, 25, 15])
labels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(values, labels=labels, startangle=90, shadow=False, autopct='%1.1f%%')
plt.show()
```
