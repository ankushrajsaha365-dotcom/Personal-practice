If k is not positive, your function should return l unchanged. Note that your function should not change itself, and should return the rotated list. A list rotation consists of taking the last element and moving it to the front. For instance, if we rotate the list [1,2,3,4,5], we get [5,1,2,3,4]. If we rotate it again, we get [4,5,1,2,3]. 
Here is an examples to show how your function should work:
Input:	rotatelist([1,2,3,4,5],3)
Output:  [3, 4, 5, 1, 2]

Write a user define Python Program to check whether a positive number is as SUM of SQUAREs or NOT. A positive integer m is a sum of squares if it can be written as k + l where k > 0, l > 0 and both k and l are perfect squares. 
Here are some examples to show how your function should work: 
        sumofsquares(25)   25 is a SUM of SQUAREs,
        sumofsquares(36)   25 is NOT a SUM of SQUAREs,
        sumofsquares(225)   25 is a SUM of SQUAREs
        

```python
# Code 1
def rotate_list(txt, rotate):
    if rotate < 0:
        return txt
    else:
        for i in range(rotate):
            txt.insert(0, txt.pop())
        return txt
x = int(input())
txt = [input() for _ in range(x)]  
rotate = int(input("Number of Rotations: "))
print(txt)
rotate_list(txt, rotate)
```

```python
# Code 2
def palindrome(s):
    if s == s[::-1]:
        print(f"{s} is a palindrome")
    else:
        print(f"{s} is not a palindrome")
s = input("Enter a string: ")
palindrome(s)
```

```python
# Code 3
def is_pangram(s):
    s = s.replace(" ", "").lower()
    alphabet = set('abcdefghijklmnopqrstuvwxyz')
    return set(s) >= alphabet

def append_middle(s1, s2):
    mid = len(s1) // 2
    return s1[:mid] + s2 + s1[mid:]

# Example usage:
pangram_str = "The quick brown fox jumps over the lazy dog"
print("Is pangram:", is_pangram(pangram_str))

s1 = "abcdef"
s2 = "123"
print("Appended string:", append_middle(s1, s2))
```

```python
# Code 4
def check(i, f, x):
    temp = 0
    for num in range(i, f + 1):
        if num == x:
            print(f"{x} is present at index {num - i}")
        else:
            temp += 1
            if temp == (f - i + 1):
                print(f"{x} is not present in the range")

i = int(input("Enter Lower Limit : "))
f = int(input("Enter Upper Limit : "))
x = int(input("Enter Number to be Searched : "))
check(i, f, x)
```

```python
# Code 5
def count(txt):
    upper, lower = 0, 0
    for i in range(len(txt)):
        if ord(txt[i]) >= 65 and ord(txt[i]) <= 90:
            upper += 1
        elif ord(txt[i]) >= 97 and ord(txt[i]) <= 122:
            lower += 1
        else:
            continue
    return upper, lower

txt = str(input("Enter a Sentence: "))
upper, lower = count(txt)
print(f"Uppercase Letters : {upper}")
print(f"Lowercase Letters : {lower}")
```

```python
# Code 6
def sum(num):
    for i in range(1000):
        for j in range(1000):
            if num == i*i + j*j:
                print(f"{num} can be expressed as the sum of squares of {i} and {j}")
                return
    print(f"{num} cannot be expressed as the sum of squares of two numbers")
num = int(input("Enter a Number: "))
sum(num)
    
```
