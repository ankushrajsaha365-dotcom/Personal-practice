6. Write a function to count vowels in a given string.
    Return the count.
    

```python
def count(str):
    count = 0
    str = str.lower()
    for i in str:
        if i == 'a' or i == 'e' or i == 'i' or i == 'o' or i == 'u':
            count += 1
    print(count)
    
statement = input()
count(statement)
```

```python
def count_vowels(s):
    c = 0
    s = s.lower()
    for ch in s:
        if ch in "aeiou":
            c += 1
    return c

statement = input()
print(count_vowels(statement))
```

7. Write a Python function that returns the factorial of a number.

Using iteration, not recursion.


```python
# using recursion
def fact1(n):
    if n == 1 or n == 0:
        return 1
    else:
        return n*fact1(n-1)

# using iteration    
def fact2(n):
    f = 1
    for i in range(1,n+1):
        f*=i
    return f



num = int(input())

print("Using Recursion",fact1(num))
print("Using Iteratin",fact2(num))
```
