```python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        self.x = x
        if x == x[::-1]:
            return True
        else:
            print(x[::-1])
            return False

x = int(input())        
sol = Solution()
check = sol.isPalindrome(str(x))
print(check)
```

```python
class Solution(object):
    def countSymmetricIntegers(self, low, high):
        count = 0
        for i in range(low,high):
            s = str(i)
            cut = len(str(i))//2
        
            if sum(map(int, s[:cut])) == sum(map(int, s[cut:])):
                count+=1
        return count   

low = int(input()) 
high = int(input())       
sol = Solution()    
check = sol.countSymmetricIntegers(low,high)
print(check)
```

```python
class Solution(object):
    def countSymmetricIntegers(self, low, high):
        count = 0
        for i in range(low, high + 1):   # include high
            s = str(i)

            # symmetric integers must have even number of digits
            if len(s) % 2 != 0:
                continue

            cut = len(s) // 2   # integer division

            if sum(map(int, s[:cut])) == sum(map(int, s[cut:])):
                count += 1

        return count


low = int(input())
high = int(input())

sol = Solution()
check = sol.countSymmetricIntegers(low, high)
print(check)
```
