```python
class Class1:
    def m(self):
        print("In Class1")

class Class2(Class1):
    def m(self):
        print("In Class2")
        Class1.m(self)
class Class3(Class1):
    def m(self):
        print("In Class3")
        Class1.m(self)

class Class4(Class2, Class3):   
    def m(self):
        print("In Class4")
        Class2.m(self)
        Class3.m(self)

obj = Class4()
obj.m()
```

```python
class Class1:
    def m(self):
        print("In Class1")

class Class2(Class1):
    def m(self):
        print("In Class2")
        super().m()
class Class3(Class1):
    def m(self):
        print("In Class3")
        super().m()

class Class4(Class2, Class3):
    def m(self):
        print("In Class4")
        super().m()

obj=Class4()
print(obj.m())
```

```python
class ABC:
    pass

from abc import ABC

class Polygon(ABC):
    def psides(self):
        pass

class Triangle(Polygon):
    def psides(self):
        print("Triangle has 3 sides")

class Rectangle(Polygon):
    def psides(self):
        print("Rectangle has 4 sides")

R = Triangle()

R.psides()

K = Rectangle()

K.psides()
```

```python
class Student:
    def __init__(self,name,a,b,c):
        self.name = name
        self.a = a
        self.b = b
        self.c = c
       
    def display(self):
        avg = (self.a + self.b + self.c)/3
        print(f"Name: {self.name} , avg marks: {avg:.2f}")
student1 = Student("Ram",98,97,92)
student2 = Student("Shyam",100,99,95)
student3 = Student("Partha",97,89,93)

student1.display()
student2.display()
student3.display()



```
