DICTIONARY

```python
sdict = {

"name": "xyz",

"roll": "123",

"year": 2019

}

print(sdict)
```

```python
x=sdict["name"]
print(x)
```

```python
x = sdict.get("year")
print(x)
```

```python
sdict["year"] = 2020
print(sdict["year"])
print(sdict)
```

```python
for x in sdict:
    print(x, sdict[x])
```

```python
for x, y in sdict.items():
    print(x, y)
if "year" in sdict:
    print("Yes, 'year' is one of the keys in the dictionary")
```
