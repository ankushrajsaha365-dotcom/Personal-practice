Write a Program that prompts user to enter two numbers
and displays their sum. Raise an exception and handle it if a
non-number value is given as input.

```python
try:
    a = int(input("Enter a: "))
    b = int(input("Enter b: "))
    print("Sum =", a + b)

except ValueError:
    print("Non-number value is given as input")
```

Write a program that accepts date of birth along with other
personal details of a person. Throw an exception if an
invalid date is entered.

```python
dob = input("Enter your date of birth(dd-mm-yyyy):")

dob = dob.split('-')

day = int(dob[0])
month = int(dob[1])
year = int(dob[2])
print(day,month,year)

from datetime import datetime

try:
    datetime(year, month , day)
    print("Valid date")
except ValueError:
    print("Invalid date")
```

Write a program that finds square root of a number. Throw
an exception if a negative number is entered.

```python
try:
    num = int(input("Enter a number: "))

    if num < 0:
        raise ValueError("Negative number entered")

    sq_root = num ** 0.5

except ValueError as e:
    print(e)

else:
    print("Square root is:", sq_root)
```

Write a Python program that prompts the user to enter a number. If the number is
positive or zero print it, otherwise raise an exception ValueError.

```python
try:
    num = int(input("Enter a number:"))

    if num < 0:
        raise ValueError("The number is negative!!!")
except ValueError as Ve:
    print(Ve)
else:
    if num > 0:
        print("The number is positive")
    else:
        print("The number is 0")
```

Write a Python program that randomly generates a number. Raise a user-defined
exception if the number is below 0.1.

```python
import random

try:
    # num = random.random()
    num = random.uniform(0,10)

    if num <= 0.1:
        raise ValueError("Number below 0.1")
except ValueError as Ve:
    print(Ve)
else:
    print(num)
```
