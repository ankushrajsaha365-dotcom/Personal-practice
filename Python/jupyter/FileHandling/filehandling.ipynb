```python
num = ['One', 'Two', 'Three', 'Four', 'Five', 'Six']

with open('test.txt', "w") as mf:
    for c in num:
        mf.write("%s\n" % c)
```

```python
c = open('test.txt')

print(c.read())
```

```python
num = ['One', 'Two', 'Three', 'Four', 'Five', 'Six']

with open('xyz.txt', "w") as mf2:
    for c in num:
        mf2.write("%s," % c)
```

```python
c = open('xyz.txt')

word=c.read()

print(word.split(','))

mf2.close()
```
