```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("popular_movie_genres.csv")

fig, ax = plt.subplots(figsize=(10, 6))

ax.grid(True, linestyle="--", alpha=0.6)


scatter = ax.scatter(
    df["Budget_Millions"],
    df["Box_Office_Millions"],
    c=df["Genre_ID"],
    cmap="viridis",
    alpha=0.7,
    s=df["Popularity_Score"] / 2,
    edgecolors="w",
    linewidth=0.5,
)

ax.set_title(
    "Movie Budget vs. Box Office Revenue by Genre",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
ax.set_xlabel("Budget (Millions USD)", fontsize=12)
ax.set_ylabel("Box Office (Millions USD)", fontsize=12)


cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label("Genre ID", fontsize=10)


plt.show()
```
