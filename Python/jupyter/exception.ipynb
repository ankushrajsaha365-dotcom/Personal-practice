<h3>When an error occurs, or exception as we call it,
Python will normally stop and generate an error
message.
These exceptions can be handled using
the try statement:

The try block lets you test a block of code for errors

The except block lets you handle the error

The finally block lets you execute code, regardless of
the result of the try- and except blocks<h3>

```python
try:
    print(x)
except:
    print("An exception occurred")
```

```python
try:
    print(x)
except NameError:
    print("Variable x is not defined")
except:
    print("Something else went wrong")
```

```python
try:
    print("Hello")
except:
    print("Something went wrong")
else:
    print("Nothing went wrong")
```

```python
try:
    print(x)
except:
    print("Something went wrong")
finally:
    print("The 'try except' is finished")
```

```python
try:
    f = open("hello.txt")
    f.write("NO WAY!!!!")
except:
    print("Something went wrong when writing to the file")
finally:
    f.close()
```

```python
(a,b) = (6,0)

try:
    # simple use of try-except block for handling errors
    g = a/b
except ZeroDivisionError:
    print ("This is a DIVIDED BY ZERO error")
```

```python
import math

x=-10

try:
    print(f'Square Root of {x} is {math.sqrt(x)}')

except ValueError as ve:
    print(f'You entered {x}, which is not a positive number.')
```

```python
a=1

while a:
    a=a+1

    if a <= 20:
        print(a)
    else:
        raise StopIteration
```

```python
try:
    a = int(input("Enter a number: "))
    print(10 / a)
except:
    print("Something went wrong!")
```

```python
try:
    a = int(input("Enter a number: "))
    print(10 / a)
except ValueError:
    print("Input must be a number!")
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

```python
try:
    a = int(input("Enter a number: "))
    result = 10 / a
except ZeroDivisionError:
    print("You cannot divide by zero.")
else:
    print("Division successful:", result)
finally:
    print("Program ended.")
```

```python
age = int(input("Enter age: "))

if age < 0:
    raise ValueError("Age cannot be negative!")
else:
    print("Valid age.")
```

```python
try:
    f = open("data.txt", "r")
    print(f.read())
except FileNotFoundError:
    print("File not found!")
```
