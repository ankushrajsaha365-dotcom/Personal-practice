```python
class Student:
    def __init__(self,name,roll):
        self.name = name
        self.roll = roll 
    def displayStudent(self):
        print(f"Name: {self.name} and Roll no. : {self.roll}")
student1 = Student("Ram",1)
student2 = Student("Shyam",2)
student3 = Student("Lakshman",3)

# print(student1.name)

student1.displayStudent()
student2.displayStudent()
student3.displayStudent()

```

```python
class Factorial:
    def __init__(self, num):
        self.num = num
        self.fact = self.CalcFact()   # calculate factorial automatically

    def CalcFact(self):
        fact = 1
        for i in range(1, self.num + 1):
            fact *= i
        return fact
    
    def displayFact(self): 
        print(f"Number: {self.num} and factorial is: {self.fact}")


n = int(input("Enter a number: "))
fact = Factorial(n)
fact.displayFact()
```

```python
class average:
    def getvalue(self):
        self.x = int(input("Enter the value of x:"))
        self.y = int(input("Enter the value of y:"))
    def putvalue(self):
        avg = (self.x +self.y)/2
        print("Average is:",avg)

avg = average()
avg.getvalue()
avg.putvalue()
```
