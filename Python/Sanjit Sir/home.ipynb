```python
class Car:
    def __init__(self,brand,powersource):
        self.brand = brand
        self.powersource = powersource

    def show(self):
        print(f"Car:{self.brand} , powersource:{self.powersource}")

car1 = Car("Tesla", "Electricity")
car2 = Car("BMW", "Petroleum")

car1.show()
car2.show()
```

```python
class student:
    def __init__(self,id,name,roll):
        self.id = id
        self.name = name
        self.roll = roll

    def show(self):
        print(f"id:{self.id}    name:{self.name}    roll:{self.roll}")

student1 = student(1,"Ram",1)
student2 = student(2,"Shyam",6)
student3 = student(3,"Jodu",3)
student4 = student(4,"Modhu",9)
student5 = student(5,"Robi",4)


student1.show()
student2.show()
student3.show()
student4.show()
student5.show()
```

```python
class Anime:
    def __init__(self,name,pop):
        self.name = name
        self.pop = pop
    def display(self):
        print(f"{self.name} is rated {self.pop}")

anime1 = Anime("One Piece",9.5)

anime2 = Anime("Naruto",9.3)

anime3 = Anime("Bleach",9.4)

anime4 = Anime("AOT",9.6)

anime1.display()
anime2.display()
anime3.display()
anime4.display()

```
