Here are some examples to show how your function should work: 
        sumofsquares(25)   25 is a SUM of SQUAREs,
        sumofsquares(36)   25 is NOT a SUM of SQUAREs,
        sumofsquares(225)   25 is a SUM of SQUAREs
        

```python
# Code 1
def rotatelist(lst, rot):
    if rot < 0:
        return lst  # Alternatively, handle left rotation here
    else:
        for i in range(rot):
            lst.insert(0, lst.pop())
        return lst

elements = []
x = int(input("Enter number of elements: "))
print("The elements are:")
for k in range(x):
    val = int(input())
    elements.append(val)
print("Actual list:",elements)

rot = int(input("Number of Rotations: "))
rotated_list = rotatelist(elements, rot)

print("Rotated list:", rotated_list)
```

```python
# Code 2
def palindrome(s):
    if s == s[::-1]:
        print(f"{s} is a palindrome")
    else:
        print(f"{s} is not a palindrome")
string = input("Enter a string: ")
palindrome(string)
```

```python
# Code 3
def is_pangram(s):
    s = s.replace("", "").lower()
    alphabet = set('abcdefghijklmnopqrstuvwxyz')
    return set(s) >= alphabet

# Example usage:
pangram_str = input("Enter a string:")
print("Is pangram:", is_pangram(pangram_str))

```

```python
def append_middle(s1, s2):
    mid = len(s1) // 2
    return s1[:mid] + s2 + s1[mid:]
a = input("Enter string a:")
b = input("Enter string b:")
print("Appended string:", append_middle(a, b))
```

```python
# Code 4
def check(i, f, x):
    temp = 0
    for num in range(i, f + 1):
        if num == x:
            print(f"{x} is present at index {num - i}")
        else:
            temp += 1
            if temp == (f - i + 1):
                print(f"{x} is not present in the range")

i = int(input("Enter Lower Limit : "))
f = int(input("Enter Upper Limit : "))
x = int(input("Enter Number to be Searched : "))
check(i, f, x)
```

```python
# Code 5
def count(txt):
    up, low = 0, 0
    for i in range(len(txt)):
        if ord(txt[i]) >= 65 and ord(txt[i]) <= 90:
            up += 1
        elif ord(txt[i]) >= 97 and ord(txt[i]) <= 122:
            low += 1
        else:
            continue
    return up , low

string = str(input("Enter a Sentence: "))
upper, lower = count(string)
print(f"Uppercase Letters : {upper}")
print(f"Lowercase Letters : {lower}")
```
